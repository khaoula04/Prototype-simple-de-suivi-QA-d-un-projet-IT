Charger le fichier CSV : 

```python
import pandas as pd 
df = pd.read_csv("tickets_demo.csv")
df.head()
```

Convertir les colonnes de date en vrai dates : 

```python
df["date_creation"] = pd.to_datetime(df["date_creation"])
df["date_limite"] = pd.to_datetime(df["date_limite"])
```

Creation de nouvelle colonne pour les tickets en retard : 

```python
from datetime import date

aujourd_hui = pd.Timestamp(date.today())
df["en_retard"] = (df["date_limite"] < aujourd_hui) & (df["statut"] != "termine")
df.head()
```

Afficher les tickets en retard : 

```python
df[df["en_retard"] == True][["ticket_id", "date_limite", "statut"]]
```

```python
total_tickets = len(df)
tickets_termines = len(df[df["statut"] == "termine"])
tickets_bloques = len(df[df["statut"] == "bloque"])
tickets_en_retard = len(df[df["en_retard"] == True])

print(f"tickets totales : {total_tickets}")
print(f"tickets termines :{tickets_termines}")
print(f"tickets bloques : {tickets_bloques}")
print(f"tickets en retard : {tickets_en_retard}")
```

Creation de colonne "sans responsable"

```python
df["sans_responsable"] = df["responsable"].isna()
df.head()
```

Creation de colonne "Critique" :

```python
df["critique"] = df["priorite"] == "critique"
df.head()
```

```python
synthese = f"""
Synthèse du projet — {date.today().strftime('%d/%m/%Y')}

Total tickets : {total_tickets}
Terminés : {tickets_termines}
Bloqués : {tickets_bloques}
En retard : {tickets_en_retard}
Sans responsable : {df['sans_responsable'].sum()}
Tickets critiques : {df['critique'].sum()}

Points d'attention principaux :
- {tickets_en_retard} ticket(s) dépassent leur date limite et nécessitent une action rapide.
- {df['sans_responsable'].sum()} ticket(s) n'ont pas de responsable assigné.
"""

print(synthese)
```
